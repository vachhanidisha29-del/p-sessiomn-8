```python
import os
import pandas as pd
import numpy as np

np.random.seed(42)

if not os.path.exists('gym_members_v1.csv'):
    base_df = pd.DataFrame({
        'member_id': range(1, 121),
        'plan': np.random.choice(['Basic', 'Standard', 'Premium'], size=120)
    })
    base_df.to_csv('gym_members_v1.csv', index=False)

gym_df = pd.read_csv('gym_members_v1.csv')

messy_names = [
    'riya SHAH',
    ' KARAN patel ',
    'Neha Joshi',
    '  arjun   MEHTA',
    'PRIYA   desai ',
    'rohan VERMA',
    ' ananya  nair',
    'VIKRAM singh ',
    'Sneha   KAPOOR',
    '  dev  sharma  '
]

messy_goals = [
    'Weight Loss',
    'weight-loss',
    'WEIGHT LOSS',
    'Muscle Gain',
    'muscle_gain',
    'Stamina'
]

gym_df['member_name'] = np.random.choice(messy_names, size=len(gym_df))
gym_df['fitness_goal'] = np.random.choice(messy_goals, size=len(gym_df))

gym_df.to_csv('gym_members_messy.csv', index=False)
```

```python
gym_df['name_clean'] = (
    gym_df['member_name']
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True)
    .str.title()
)

print(gym_df['name_clean'].unique())
assert gym_df['name_clean'].nunique() <= 10
```

```python
gym_df['goal_clean'] = (
    gym_df['fitness_goal']
    .str.lower()
    .str.replace(r'[-_]', ' ', regex=True)
)

weight_loss_count = gym_df['goal_clean'].str.contains('weight').sum()
print(weight_loss_count)

goal_counts = gym_df['goal_clean'].value_counts()
print(goal_counts)
assert len(goal_counts) == 3
```

```python
gym_df[['first_name', 'last_name']] = gym_df['name_clean'].str.split(' ', expand=True)
gym_df = gym_df.drop(columns=['member_name', 'fitness_goal'])

gym_df.to_csv('gym_members_v2.csv', index=False)
print(gym_df.head())
```
